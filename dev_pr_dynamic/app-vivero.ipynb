## 1) Descripción del problema **[1 pts]**
El objetivo es moldear y calcular la produccion de nuevas plantas en un vivero durante el mes **n** que pasa de manera controlada. Se observa que esta sigue un crecimiento controlado en el cual en el mes **0** (n = 1) no hay produccion de plantas, en el mes **1** (n = 1) hay producción de solo una planta y a partir del mes 2 (n >= 2) la produccion de plantas equivale a a la suma de de la produccion de los meses inmediatos anteriores **T(n)=T(n-1)+T(n-2)**.

### 5) Análisis de complejidad **[15 pts]**

**Análisis de complejidad**

**Justificación:**

**Sin optimización:** El código repite cálculos de meses anteriores y se vuelve extremadamente lento (O(2^n)).
**Con optimización:** Guarda los resultados en memoria. Así calcula cada mes una sola vez y el proceso pasa a ser directo y rápido (O(n)).

**Función T(n):**

**Casos base:** T(0) = 1, T(1) = 1
**Caso general:** T(n) = T(n-1) + T(n-2) + 1
**Complejidad temporal con memoización:** O(n)
**Complejidad espacial:** O(n) por la tabla de memoria y la pila de llamadas recursivas.

**Mejor y peor caso:**
Mejor caso: Ω(1) cuando el mes n ya está guardado en memoria o es un caso base.
Peor caso: O(n) cuando la memoria está vacía y se deben calcular todos los n meses previos.

## 6) Código funcional **[15 pts]**
Incluye tu implementación completa.

No pegues código incompleto; debe ser ejecutable.


```python
def calcular_plantas(n, memoria={0: 0, 1: 1}):

    if n in memoria:

        return memoria[n]

    memoria[n] = calcular_plantas(n - 1, memoria) + calcular_plantas(n - 2, memoria)

    return memoria[n]

while True:

    entrada = input("\nIngrese el número de mes (o 'salir' para terminar): ")

    if entrada.lower() == 'salir':

        print("Programa finalizado.")

        break

    try:

        mes = int(entrada)

        if mes < 0:

            print("ERROR: Ingresó un número negativo. Intente de nuevo con un número positivo.")

        else:

            resultado = calcular_plantas(mes)

            print(f"Plantas producidas en el mes {mes}: {resultado}")

    except ValueError:

        print("ERROR: Ingresó letras o caracteres no válidos. Debe ingresar solo números enteros.")
 
```
