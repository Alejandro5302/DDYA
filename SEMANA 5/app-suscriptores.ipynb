## 1) Descripción del problema **[1 pts]**

Una plataforma de streaming está analizando el crecimiento de nuevos suscriptores.
Se ha observado que el número de nuevos suscriptores en el mes n sigue el siguiente comportamiento:

- En el mes 0 se registraron 0 nuevos usuarios.
- En el mes 1 se registró 1 nuevo usuario.
- A partir del mes 2, el número de nuevos usuarios es igual a la suma de los nuevos usuarios de los dos meses anteriores, debido al efecto combinado de referidos del mes anterior y recomendaciones acumuladas del mes previo.

**Relación de recurrencia:**

T(n) = 0, si n = 0

T(n) = 1, si n = 1

T(n) = T(n-1) + T(n-2), si n ≥ 2

Se busca determinar, usando memoización, cuántos nuevos suscriptores habrá en el mes n.

## 7) Análisis de complejidad **[5 pts]**

**Función T(n) - versión recursiva sin memoización**

T(n) = T(n-1) + T(n-2) + O(1), con T(0) = O(1) y T(1) = O(1)

Esta recurrencia, al igual que la de Fibonacci clásico, tiene solución **O(2^n)** en tiempo (crecimiento exponencial), ya que cada llamada genera dos llamadas recursivas y el árbol de recursión no reutiliza los subproblemas ya resueltos.

**Función T(n) - versión con memoización**

Al guardar en una tabla (o diccionario) los resultados de T(0), T(1), ..., T(n) ya calculados, cada subproblema se resuelve una sola vez:

T(n) = O(n) en tiempo, ya que se calcula cada valor de 0 a n exactamente una vez con trabajo O(1) por valor.

El espacio usado es O(n) por la tabla de memoización (o O(1) si se usa la versión iterativa que solo guarda los dos últimos valores).

**Mejor y peor caso**

- Sin memoización: el mejor y el peor caso son ambos O(2^n) en función de n, ya que la recurrencia siempre se expande igual sin importar el valor concreto de n (no hay una entrada "fácil" salvo n=0 o n=1).
- Con memoización: el mejor y peor caso son ambos O(n), pues siempre se recorre la tabla desde 0 hasta n.

*(Agrega aquí tu gráfica comparando el crecimiento teórico O(2^n) vs O(n), o usa la gráfica generada en la sección de Pruebas de Desempeño más abajo.)*


```python
def suscriptores_recursivo(n):
    if n == 0:
        return 0
    if n == 1:
        return 1
    return suscriptores_recursivo(n - 1) + suscriptores_recursivo(n - 2)


def suscriptores_memo(n):
   
    if n < 0:
        raise ValueError("n debe ser un entero no negativo")
    if n == 0:
        return 0
    if n == 1:
        return 1
    tabla = [0] * (n + 1)
    tabla[1] = 1
    for i in range(2, n + 1):
        tabla[i] = tabla[i - 1] + tabla[i - 2]
    return tabla[n]


try:
    mes = int(input("Ingrese el mes (n) para calcular los nuevos suscriptores: "))
    if mes < 0:
        print("Por favor ingrese un valor entero no negativo.")
    else:
        print("Mes consultado:", mes)
        resultado = suscriptores_memo(mes)
        print("Nuevos suscriptores en el mes", mes, "->", resultado)
except ValueError:
    print("Por favor ingrese un valor numérico entero.")
```

Medición empírica del tiempo de ejecución, gráfica de comportamiento y relación con el análisis de complejidad teórico.

#### Código base para benchmarking


```python

def bench(n: int, step: int = 1) -> list[tuple[int, float]]:
    print("Generando datos ...", end='')
    import time
    results: list[tuple[int, float]] = []

    for size in range(1, n, step):
        ti = time.perf_counter()
        output = suscriptores_memo(size)
        tf = time.perf_counter()

        delta = (tf - ti) * 1000
        print(".", end="")
        results.append((size, delta))
    return results




def dibujar(x, y, dimensiones=(7, 5)):
    assert len(x) == len(y)
    import matplotlib.pyplot as plt
    plt.figure(figsize=dimensiones)
    plt.plot(x, y)
    plt.xlabel("n (mes)")
    plt.ylabel("tiempo (ms)")
    plt.title("Tiempo de ejecución de suscriptores_memo(n)")
    plt.show()

tiempos = bench(2000, step=100)

x = [size for size, timing in tiempos]
y = [timing for size, timing in tiempos]

dibujar(x, y)
```

Si se probara la versión `suscriptores_recursivo` (sin memoización) con valores de n mayores a ~30, el tiempo crecería de forma exponencial (O(2^n)), evidenciando la diferencia frente a la versión optimizada.
